```python
try:
    balance = float(input("Enter Cash Book Balance: "))

    transactions = []

    n = int(input("Enter number of transactions: "))

    for i in range(n):
        print("\nTransaction", i + 1)
        try:
            amount = float(input("Enter amount: "))
            type = input("Enter Debit or Credit: ")

            transactions.append([amount, type])
        except ValueError:
            print("Invalid Input , Enter a Valid Input")

    for transaction in transactions:

        amount = transaction[0]
        type = transaction[1]

        if type.lower() == "credit":
            balance = balance + amount

        elif type.lower() == "debit":
            balance = balance - amount

    print("\nMONTHLY BANK RECONCILIATION STATEMENT")
    print("--------------------------------------")
    print("Final Reconciled Balance:", balance)


except ValueError:
    print("Invalid Input , Enter a Valid Input")


```
