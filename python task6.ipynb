```python
import pandas as pd
df = pd.read_csv("financial_dataset.csv")

print("Original Dataset")
print(df.head())

print("\nNumber of rows and columns:")
print(df.shape)
```

```python
print("\nColumn Names:")
print(df.columns)



```

```python
print("\nData Types:")
print(df.dtypes)

```

```python
print("\nBasic Statistics:")
print(df.describe())
```

```python
print("\nMissing Values in Each Column:")
print(df.isnull().sum())
```

```python

df["Sales"] = df["Sales"].fillna(df["Sales"].median())

df["Profit"] = df["Profit"].fillna(df["Profit"].median())

df["Category"] = df["Category"].fillna(df["Category"].mode()[0])


df["City"] = df["City"].fillna(df["City"].mode()[0])


print("\nMissing Values After Filling:")
print(df.isnull().sum())
```

```python
print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())


```

```python
df = df.drop_duplicates()
```

```python
print("\nCategory Values Before Cleaning:")
print(df["Category"].value_counts())
```

```python
# Remove extra spaces
df["Category"] = df["Category"].str.strip()

# Convert first letter of each word to uppercase
df["Category"] = df["Category"].str.title()


print("\nCategory Values After Cleaning:")
print(df["Category"].value_counts())
```

```python
print("\nPayment Status Before Cleaning:")
print(df["Payment_Status"].value_counts())
```

```python
df["Payment_Status"] = df["Payment_Status"].str.strip()
df["Payment_Status"] = df["Payment_Status"].str.title()


print("\nPayment Status After Cleaning:")
print(df["Payment_Status"].value_counts())
```

```python
print("\nData Types Before Conversion:")
print(df.dtypes)
```

```python
df["Sales"] = pd.to_numeric(
    df["Sales"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)


print("\nData Types After Conversion:")
print(df.dtypes)
```

```python
print("\nCleaned Dataset:")
print(df.head(20))
```

```python
df.to_csv(
    "financial_dataset_2000_cleaned_by_student.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
```
