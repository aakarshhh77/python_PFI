```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("financial_statements_dataset.csv")
```

```python
df
```

```python
df["Net_Income"] = (
    df["Revenue"]
    - df["Cost_of_Goods_Sold"]
    - df["Operating_Expenses"]
    - df["Interest_Expense"]
    - df["Tax"]
)
```

```python
df["Current_Ratio"] = (
    df["Current_Assets"] / df["Current_Liabilities"]
)
```

```python
df["ROE_Percent"] = (
    df["Net_Income"] / df["Equity"]
) * 100
```

```python


df["EPS"] = (
    (df["Net_Income"] - df["Preferred_Dividends"])
    / df["Shares_Outstanding"]
)
```

```python
print("\nFinancial Ratio Analysis:")
print(
    df[
        ["Year", "Net_Income",
         "Current_Ratio", "ROE_Percent", "EPS"]
    ]
)
```

```python
print("\nHighest Current Ratio:")
print(df.loc[df["Current_Ratio"].idxmax(), ["Year", "Current_Ratio"]])

print("\nHighest ROE:")
print(df.loc[df["ROE_Percent"].idxmax(), ["Year", "ROE_Percent"]])

print("\nHighest EPS:")
print(df.loc[df["EPS"].idxmax(), ["Year", "EPS"]])
```

```python

plt.figure(figsize=(8, 5))
plt.plot(df["Year"], df["EPS"], marker="o")
plt.title("EPS Trend")
plt.xlabel("Year")
plt.ylabel("EPS (₹)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["Year"], df["ROE_Percent"], marker="o")
plt.title("ROE Trend")
plt.xlabel("Year")
plt.ylabel("ROE (%)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["Year"], df["Current_Ratio"], marker="o")
plt.title("Current Ratio Trend")
plt.xlabel("Year")
plt.ylabel("Current Ratio")
plt.grid(True)
plt.show()

```
